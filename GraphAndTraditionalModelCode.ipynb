```python
import os
import subprocess
import re

def extract_function_cfgs_to_dot(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)

    for filename in os.listdir(input_dir):
        if filename.endswith(".sol"):
            sol_path = os.path.join(input_dir, filename)
            contract_name = os.path.splitext(filename)[0]

            print(f"Processing: {sol_path}")

            try:
                # Run Slither with `--print cfg` and capture output
                result = subprocess.run(
                    ["slither", sol_path, "--print", "cfg"],
                    stdout=subprocess.PIPE,
                    stderr=subprocess.PIPE,
                    text=True,
                    check=True
                )

                output = result.stdout
                # Split output into function blocks
                blocks = re.split(r"CFG for function '(.+?)':", output)

                # blocks[0] is the header before any CFGs
                for i in range(1, len(blocks), 2):
                    func_name = blocks[i].strip().replace(" ", "_")
                    dot_body = blocks[i + 1].strip()

                    # Format as a DOT file
                    dot_graph = f"digraph \"{func_name}\" {{\n{dot_body}\n}}"

                    dot_filename = f"{contract_name}_{func_name}.dot"
                    dot_path = os.path.join(output_dir, dot_filename)

                    with open(dot_path, "w") as f:
                        f.write(dot_graph)

            except subprocess.CalledProcessError as e:
                print(f"Error processing {filename}: {e.stderr}")

# Example usage
if __name__ == "__main__":
    input_contracts_folder = "ESC_SOL"
    output_dot_cfg_folder = "CFG_DOT_FILES"
    extract_function_cfgs_to_dot(input_contracts_folder, output_dot_cfg_folder)
```

```python
import os
import shutil

def transfer_dot_files(source_folder, destination_folder):
    """
    Transfer all .dot files from source_folder to destination_folder.
    
    Args:
        source_folder (str): Path to the source folder
        destination_folder (str): Path to the destination folder
    """
    # Create destination folder if it doesn't exist
    os.makedirs(destination_folder, exist_ok=True)
    
    # Get all .dot files in the source folder
    dot_files = [f for f in os.listdir(source_folder) if f.endswith('.dot')]
    
    if not dot_files:
        print(f"No .dot files found in {source_folder}")
        return
    
    # Transfer each file
    for file in dot_files:
        source_path = os.path.join(source_folder, file)
        dest_path = os.path.join(destination_folder, file)
        
        shutil.move(source_path, dest_path)
        print(f"Moved: {file}")
    
    print(f"\nSuccessfully transferred {len(dot_files)} .dot files to {destination_folder}")

if __name__ == "__main__":
    # Get folder paths from user
    source = input("Enter source folder path: ").strip()
    destination = input("Enter destination folder path: ").strip()
    
    # Validate paths
    if not os.path.isdir(source):
        print(f"Error: Source folder '{source}' does not exist")
    else:
        transfer_dot_files(source, destination)
```

```python
import os
import json
import pydot
from pathlib import Path

def dot_to_json(dot_file_path, json_file_path):
    """Convert a .dot file to .json format"""
    try:
        # Parse the DOT file
        graphs = pydot.graph_from_dot_file(dot_file_path)
        
        # We assume the first graph is the one we want (CFG)
        graph = graphs[0]
        
        # Convert to a dictionary structure
        graph_dict = {
            'nodes': [],
            'edges': []
        }
        
        # Process nodes
        for node in graph.get_nodes():
            node_dict = {
                'name': node.get_name(),
                'label': node.get_label(),
                'attributes': node.get_attributes()
            }
            graph_dict['nodes'].append(node_dict)
        
        # Process edges
        for edge in graph.get_edges():
            edge_dict = {
                'source': edge.get_source(),
                'destination': edge.get_destination(),
                'label': edge.get_label(),
                'attributes': edge.get_attributes()
            }
            graph_dict['edges'].append(edge_dict)
        
        # Save as JSON
        with open(json_file_path, 'w') as json_file:
            json.dump(graph_dict, json_file, indent=2)
            
    except Exception as e:
        print(f"Error processing {dot_file_path}: {str(e)}")

def convert_folder(source_folder, dest_folder):
    """Convert all .dot files in source_folder to .json in dest_folder"""
    # Create destination folder if it doesn't exist
    Path(dest_folder).mkdir(parents=True, exist_ok=True)
    
    # Process each .dot file
    for filename in os.listdir(source_folder):
        if filename.endswith('.dot'):
            source_path = os.path.join(source_folder, filename)
            dest_path = os.path.join(dest_folder, f"{Path(filename).stem}.json")
            dot_to_json(source_path, dest_path)
            print(f"Converted {filename} to {Path(dest_path).name}")

if __name__ == "__main__":
    source_folder = input("Enter source folder path (containing .dot files): ").strip()
    dest_folder = input("Enter destination folder path (for .json files): ").strip()
    
    convert_folder(source_folder, dest_folder)
    print("Conversion complete!")
```

```python
import os
import json
import torch
import torch.nn.functional as F
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GINConv, global_add_pool
import pandas as pd
from torch.nn import Sequential, Linear, ReLU

# === Configuration ===
INPUT_FOLDER = "CFG_JSON_FILES"  # <-- Replace with your folder path
OUTPUT_CSV = "esc_gin_emb.csv"
EMBED_DIM = 16
BATCH_SIZE = 32
EPOCHS = 30


# === Load CFGs as PyG Graphs ===
def load_graphs_from_folder(folder):
    data_list = []
    filenames = []

    for filename in os.listdir(folder):
        if not filename.endswith(".json"):
            continue

        try:
            with open(os.path.join(folder, filename), 'r') as f:
                cfg = json.load(f)

            # Step 1: Nodes
            nodes = cfg.get("nodes", [])
            if not nodes:
                # Create one dummy node
                nodes = [0]
            if isinstance(nodes[0], dict) and "id" in nodes[0]:
                node_ids = [n["id"] for n in nodes]
            elif isinstance(nodes[0], int):
                node_ids = nodes
            else:
                node_ids = list(range(len(nodes)))

            id_map = {nid: idx for idx, nid in enumerate(sorted(set(node_ids)))}
            num_nodes = len(id_map)
            x = torch.eye(num_nodes, dtype=torch.float)

            # Step 2: Edges
            edges_raw = cfg.get("edges", [])
            edges = []
            for e in edges_raw:
                if isinstance(e, list) and len(e) == 2:
                    src, dst = e
                    if src in id_map and dst in id_map:
                        edges.append([id_map[src], id_map[dst]])

            if not edges:
                edge_index = torch.empty((2, 0), dtype=torch.long)  # No edges
            else:
                edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()

            # Step 3: Build Graph
            data = Data(x=x, edge_index=edge_index)
            data_list.append(data)
            filenames.append(filename)

        except Exception as e:
            print(f"Skipping {filename} due to error: {e}")
            continue

    return data_list, filenames


# === GIN Model ===
class GIN(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.conv1 = GINConv(Sequential(Linear(input_dim, hidden_dim), ReLU(), Linear(hidden_dim, hidden_dim)))
        self.conv2 = GINConv(Sequential(Linear(hidden_dim, hidden_dim), ReLU(), Linear(hidden_dim, hidden_dim)))
        self.out = Linear(hidden_dim, output_dim)

    def forward(self, x, edge_index, batch):
        x = F.relu(self.conv1(x, edge_index))
        x = self.conv2(x, edge_index)
        x = global_add_pool(x, batch)
        return self.out(x)


# === Main Pipeline ===
def main():
    data_list, filenames = load_graphs_from_folder(INPUT_FOLDER)

    if not data_list:
        print("❌ No valid graphs found.")
        return

    input_dim = max([d.x.size(1) for d in data_list])
    for d in data_list:
        if d.x.size(1) < input_dim:
            padding = input_dim - d.x.size(1)
            d.x = torch.cat([d.x, torch.zeros(d.num_nodes, padding)], dim=1)

    loader = DataLoader(data_list, batch_size=BATCH_SIZE, shuffle=True)

    model = GIN(input_dim=input_dim, hidden_dim=32, output_dim=EMBED_DIM)
    optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

    # === Training ===
    model.train()
    for epoch in range(EPOCHS):
        total_loss = 0
        for batch in loader:
            optimizer.zero_grad()
            out = model(batch.x, batch.edge_index, batch.batch)
            loss = out.norm(p=2)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        print(f"Epoch {epoch+1}/{EPOCHS}, Loss: {total_loss:.4f}")

    # === Embedding Extraction ===
    model.eval()
    embeddings = []
    with torch.no_grad():
        for data in data_list:
            data.batch = torch.zeros(data.num_nodes, dtype=torch.long)
            emb = model(data.x, data.edge_index, data.batch)
            embeddings.append(emb.squeeze().numpy())

    df = pd.DataFrame(embeddings, index=filenames)
    df.to_csv(OUTPUT_CSV)
    print(f"✅ Embeddings saved to: {OUTPUT_CSV}")


if __name__ == "__main__":
    main()
```

```python
import os
import json
import torch
import torch.nn.functional as F
import pandas as pd
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import GCNConv, global_mean_pool

FEATURE_DIM = 16  # Set a fixed feature dimension

# GCN Model
def create_gcn():
    class GCN(torch.nn.Module):
        def __init__(self, input_dim=FEATURE_DIM, hidden_dim=32, output_dim=16):
            super(GCN, self).__init__()
            self.conv1 = GCNConv(input_dim, hidden_dim)
            self.conv2 = GCNConv(hidden_dim, output_dim)

        def forward(self, x, edge_index, batch):
            x = F.relu(self.conv1(x, edge_index))
            x = self.conv2(x, edge_index)
            return global_mean_pool(x, batch)  # Pooling to get graph embedding
    
    return GCN

# Function to load CFGs from JSON files and convert them to PyG format
def load_graphs_from_folder(folder):
    data_list = []
    for filename in os.listdir(folder):
        if filename.endswith(".json"):
            with open(os.path.join(folder, filename), 'r') as f:
                graph_data = json.load(f)
            
            nodes = graph_data.get("nodes", [])  # Ensure "nodes" exist
            edges = graph_data.get("edges", [])  # Ensure "edges" exist
            
            if isinstance(edges, dict):
                edges = [(int(src), int(dst)) for src, dst_list in edges.items() for dst in dst_list]
            elif not isinstance(edges, list):
                print(f"Warning: {filename} has invalid edges (not a list). Using empty list.")
                edges = []
            
            edges = [(int(e[0]), int(e[1])) for e in edges if isinstance(e, (list, tuple)) and len(e) == 2]
            
            if not edges:
                print(f"Warning: {filename} has an empty edge list but will still be processed.")
            
            x = torch.ones(len(nodes), FEATURE_DIM)  # Fixed-size node features
            edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous() if edges else torch.empty((2, 0), dtype=torch.long)
            
            data = Data(x=x, edge_index=edge_index)
            data_list.append((filename, data))
    
    return data_list

# Function to generate embeddings
def generate_embeddings(model, data_list, output_csv):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)
    model.eval()
    
    embeddings = []
    
    with torch.no_grad():
        for filename, data in data_list:
            data = data.to(device)
            embedding = model(data.x, data.edge_index, torch.zeros(data.x.size(0), dtype=torch.long, device=device))
            embeddings.append([filename] + embedding.cpu().tolist())
    
    df = pd.DataFrame(embeddings)
    df.to_csv(output_csv, index=False, header=False, sep='\t')  # Column-separated output
    print(f"Embeddings saved to {output_csv}")

# Main execution function
def main():
    folder = "CFG_JSON_FILES"  # Change this to your CFG folder path
    output_csv = "esc_gcn_emb.csv"
    
    data_list = load_graphs_from_folder(folder)
    
    if not data_list:
        print("No valid CFGs found.")
        return
    
    GCN = create_gcn()
    model = GCN()
    
    generate_embeddings(model, data_list, output_csv)

if __name__ == "__main__":
    main()
```

```python
import os
import json
import torch
import torch.nn.functional as F
from torch_geometric.data import Data, DataLoader
from torch_geometric.nn import SAGEConv
import csv

class GraphSAGE(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super(GraphSAGE, self).__init__()
        self.conv1 = SAGEConv(in_channels, hidden_channels)
        self.conv2 = SAGEConv(hidden_channels, out_channels)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return x

def load_cfg_graphs(folder_path):
    data_list = []
    filenames = []

    for file in os.listdir(folder_path):
        if not file.endswith(".json"):
            continue
        with open(os.path.join(folder_path, file), "r") as f:
            cfg = json.load(f)

        if "nodes" not in cfg or "edges" not in cfg:
            print(f"Skipping {file}: Missing required keys")
            continue

        try:
            num_nodes = len(cfg["nodes"])

            if num_nodes == 0:
                # Add 1 dummy node with zero feature
                x = torch.zeros(1, 16, dtype=torch.float)
                edge_index = torch.empty((2, 0), dtype=torch.long)
            else:
                x = torch.randn(num_nodes, 16, dtype=torch.float)

                valid_nodes = set(range(num_nodes))
                edges = cfg["edges"]
                filtered_edges = [[int(e["from"]), int(e["to"])] for e in edges
                                  if isinstance(e, dict) and "from" in e and "to" in e and
                                  int(e["from"]) in valid_nodes and int(e["to"]) in valid_nodes]
                edge_index = torch.tensor(filtered_edges, dtype=torch.long).t().contiguous() \
                             if filtered_edges else torch.empty((2, 0), dtype=torch.long)

            data = Data(x=x, edge_index=edge_index)
            data_list.append(data)
            filenames.append(file)

        except Exception as e:
            print(f"Skipping {file}: {str(e)}")
            continue

    return data_list, filenames

def generate_embeddings(input_folder, output_csv):
    dataset, filenames = load_cfg_graphs(input_folder)
    if len(dataset) == 0:
        print("No valid CFG graphs found.")
        return

    dataloader = DataLoader(dataset, batch_size=1, shuffle=False)

    model = GraphSAGE(in_channels=16, hidden_channels=32, out_channels=16)
    model.eval()

    embeddings = []

    with torch.no_grad():
        for i, data in enumerate(dataloader):
            if data.edge_index.dim() != 2:
                print(f"Skipping {filenames[i]}: Invalid edge_index shape {data.edge_index.shape}")
                continue

            emb = model(data.x, data.edge_index).mean(dim=0).tolist()
            embeddings.append(emb)

    with open(output_csv, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["Filename"] + [f"Dim{i}" for i in range(16)])
        for i in range(len(embeddings)):
            writer.writerow([filenames[i]] + embeddings[i])

    print(f"Embeddings saved to {output_csv}")

# Run the pipeline
input_folder = "CFG_JSON_FILES"
output_csv = "esc_sage_emb.csv"
generate_embeddings(input_folder, output_csv)
```

```python
import os
import json
import torch
import torch.nn.functional as F
import networkx as nx
import pandas as pd
from torch_geometric.data import Data, Dataset, DataLoader
from torch_geometric.nn import GATConv

class CFGDataset(Dataset):
    def __init__(self, folder_path, fallback_dim=10):
        self.folder_path = folder_path
        self.file_list = [f for f in os.listdir(folder_path) if f.endswith(".json")]
        self.fallback_dim = fallback_dim
        super().__init__()

    def len(self):
        return len(self.file_list)

    def get(self, idx):
        file_path = os.path.join(self.folder_path, self.file_list[idx])
        
        try:
            with open(file_path, "r") as f:
                cfg_data = json.load(f)
        except json.JSONDecodeError:
            print(f"Error: Invalid JSON in {file_path}")
            return None

        nodes = cfg_data.get("nodes", [])
        edges = cfg_data.get("edges", [])

        node_features = {}
        for i, node in enumerate(nodes):
            node_id = node.get("id", i)
            feature = node.get("feature")

            try:
                node_id = int(node_id)
                if isinstance(feature, list) and all(isinstance(x, (int, float)) for x in feature):
                    feature_tensor = torch.tensor(feature, dtype=torch.float)
                else:
                    feature_tensor = torch.zeros(self.fallback_dim, dtype=torch.float)
                node_features[node_id] = feature_tensor
            except Exception:
                node_features[node_id] = torch.zeros(self.fallback_dim, dtype=torch.float)

        if not node_features:
            node_features[0] = torch.zeros(self.fallback_dim, dtype=torch.float)

        edge_index_list = []
        for edge in edges:
            try:
                src = int(edge["source"])
                tgt = int(edge["target"])
                if src in node_features and tgt in node_features:
                    edge_index_list.append([src, tgt])
            except Exception:
                continue

        x = torch.stack([node_features[n] for n in sorted(node_features.keys())])
        edge_index = (
            torch.tensor(edge_index_list, dtype=torch.long).t().contiguous()
            if edge_index_list else torch.empty((2, 0), dtype=torch.long)
        )

        return Data(x=x, edge_index=edge_index)

class GAT(torch.nn.Module):
    def __init__(self, in_dim, hidden_dim=32, out_dim=16, heads=4):
        super().__init__()
        self.conv1 = GATConv(in_dim, hidden_dim, heads=heads)
        self.conv2 = GATConv(hidden_dim * heads, out_dim, heads=1)

    def forward(self, x, edge_index):
        x = F.elu(self.conv1(x, edge_index))
        x = self.conv2(x, edge_index)
        return x.mean(dim=0)

# Load dataset
data_folder = "CFG_JSON_FILES"
fallback_dim = 10
dataset = CFGDataset(data_folder, fallback_dim=fallback_dim)

# Use first valid sample or force fallback dimension
valid_sample = next((dataset.get(i) for i in range(len(dataset)) if dataset.get(i) is not None), None)
if valid_sample is None:
    print("No valid graphs found, using dummy input for model init")
    dummy_input = torch.zeros((1, fallback_dim))
    dummy_edge_index = torch.empty((2, 0), dtype=torch.long)
    valid_sample = Data(x=dummy_input, edge_index=dummy_edge_index)

# Initialize model
input_dim = valid_sample.x.shape[1]
gat_model = GAT(input_dim)
gat_model.eval()

# Generate embeddings
embeddings = []
file_names = []

with torch.no_grad():
    for i in range(len(dataset)):
        data = dataset.get(i)
        if data is None:
            continue
        embedding = gat_model(data.x, data.edge_index)
        embeddings.append(embedding.numpy())
        file_names.append(dataset.file_list[i])

# Save to CSV
if embeddings:
    embeddings_df = pd.DataFrame(embeddings, index=file_names)
    embeddings_df.to_csv("esc_gat_emb.csv", sep=";")
    print("Embeddings saved to gat_embeddings.csv")
else:
    print("No embeddings generated.")
```

Convert Source Code to ByteCode 

```python
import os
from solcx import compile_files, install_solc

def clean_output_folder(output_folder):
    """Remove all .txt files from output folder"""
    for filename in os.listdir(output_folder):
        if filename.endswith('.txt'):
            try:
                os.remove(os.path.join(output_folder, filename))
                print(f"Removed {filename} from output folder")
            except Exception as e:
                print(f"Error removing {filename}: {str(e)}")

def convert_solidity_to_bytecode(input_folder, output_folder, solc_version='0.8.0'):
    """
    Convert Solidity files to bytecode and save as .bin files
    
    Args:
        input_folder: Folder containing .sol files
        output_folder: Folder to store .bin files
        solc_version: Solidity compiler version
    """
    # Create output folder if needed
    os.makedirs(output_folder, exist_ok=True)
    
    # Clean existing .txt files
    clean_output_folder(output_folder)
    
    # Install compiler if needed
    try:
        install_solc(solc_version)
    except Exception as e:
        print(f"Compiler installation failed: {str(e)}")
        return

    # Process each Solidity file
    success = 0
    for filename in os.listdir(input_folder):
        if not filename.endswith('.sol'):
            continue
            
        input_path = os.path.join(input_folder, filename)
        output_path = os.path.join(output_folder, f"{os.path.splitext(filename)[0]}.bin")
        
        try:
            # Compile to bytecode only
            compiled = compile_files(
                [input_path],
                output_values=['bin'],
                solc_version=solc_version
            )
            
            # Save bytecode
            with open(output_path, 'w') as f:
                f.write(list(compiled.values())[0]['bin'])
            
            print(f"✓ {filename}")
            success += 1
            
        except Exception as e:
            print(f"✗ {filename} - {str(e)}")
    
    print(f"\nCompilation complete. Successfully compiled {success} files")

if __name__ == "__main__":
    INPUT_FOLDER = "./ESC_SOL"
    OUTPUT_FOLDER = "./ESC_BYTECODE"
    SOLC_VERSION = '0.4.25'  # Change to your required version
    
    convert_solidity_to_bytecode(INPUT_FOLDER, OUTPUT_FOLDER, SOLC_VERSION)
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, Concatenate

def load_contract_data(source_folder, bytecode_folder, labels_file):
    df = pd.read_csv(labels_file)
    df.columns = [col.strip().lower() for col in df.columns]

    if 'filename' not in df.columns:
        raise ValueError("CSV must contain a 'filename' column.")
    if 'reentrancy' not in df.columns:
        raise ValueError("CSV must contain a 'reentrancy' column for labels.")

    sources, bytecodes, labels = [], [], []

    for _, row in df.iterrows():
        fname = str(row['filename']).strip()
        label = int(row['reentrancy'])

        # Ensure proper filename extensions
        if not fname.endswith('.sol'):
            fname += '.sol'

        # Extract base for .bin file
        basename = os.path.splitext(os.path.basename(fname))[0]

        source_path = os.path.join(source_folder, fname)
        byte_path = os.path.join(bytecode_folder, f"{basename}.bin")

        if not os.path.isfile(source_path):
            print(f"⚠️ Missing source file: {source_path}")
            continue
        if not os.path.isfile(byte_path):
            print(f"⚠️ Missing bytecode file: {byte_path}")
            continue

        try:
            with open(source_path, 'r', encoding='utf-8') as f:
                source = f.read()

            with open(byte_path, 'r', encoding='utf-8') as f:
                bytecode_hex = f.read().strip()
                bytecode_tokens = ' '.join([bytecode_hex[i:i+2] for i in range(0, len(bytecode_hex), 2)])

            sources.append(source)
            bytecodes.append(bytecode_tokens)
            labels.append(label)
        except Exception as e:
            print(f"❌ Error reading {fname}: {e}")
            continue

    if len(sources) == 0:
        raise ValueError("No valid contracts loaded. Check file paths and naming consistency.")

    print(f"✅ Successfully loaded {len(sources)} contracts.")
    return sources, bytecodes, np.array(labels)

def tokenize_inputs(source_texts, bytecode_texts, max_words=10000, max_len=500):
    source_tokenizer = Tokenizer(num_words=max_words, oov_token="<OOV>")
    bytecode_tokenizer = Tokenizer(num_words=max_words, oov_token="<OOV>")

    source_tokenizer.fit_on_texts(source_texts)
    bytecode_tokenizer.fit_on_texts(bytecode_texts)

    source_seq = source_tokenizer.texts_to_sequences(source_texts)
    bytecode_seq = bytecode_tokenizer.texts_to_sequences(bytecode_texts)

    source_seq = pad_sequences(source_seq, maxlen=max_len, padding='post')
    bytecode_seq = pad_sequences(bytecode_seq, maxlen=max_len, padding='post')

    return source_seq, bytecode_seq, source_tokenizer, bytecode_tokenizer

def build_model(vocab_size_source, vocab_size_bytecode, embedding_dim=64, input_len=500):
    source_input = Input(shape=(input_len,), name='source_input')
    bytecode_input = Input(shape=(input_len,), name='bytecode_input')

    source_embed = Embedding(vocab_size_source, embedding_dim)(source_input)
    bytecode_embed = Embedding(vocab_size_bytecode, embedding_dim)(bytecode_input)

    source_lstm = LSTM(64)(source_embed)
    bytecode_lstm = LSTM(64)(bytecode_embed)

    merged = Concatenate()([source_lstm, bytecode_lstm])
    dense = Dense(64, activation='relu')(merged)
    output = Dense(1, activation='sigmoid')(dense)

    model = Model(inputs=[source_input, bytecode_input], outputs=output)
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

def main():
    SOURCE_FOLDER = 'ESC_SOL'
    BYTECODE_FOLDER = 'ESC_BYTECODE'
    LABELS_FILE = 'labelled_ESC_1001.csv'

    print("🔄 Loading contracts and labels...")
    sources, bytecodes, labels = load_contract_data(SOURCE_FOLDER, BYTECODE_FOLDER, LABELS_FILE)

    print("🔄 Tokenizing and padding sequences...")
    source_seq, bytecode_seq, source_tokenizer, bytecode_tokenizer = tokenize_inputs(sources, bytecodes)

    print("🔄 Splitting into train/test...")
    X_source_train, X_source_test, X_bytecode_train, X_bytecode_test, y_train, y_test = train_test_split(
        source_seq, bytecode_seq, labels, test_size=0.2, random_state=42
    )

    print("🔄 Building the model...")
    model = build_model(
        vocab_size_source=len(source_tokenizer.word_index) + 1,
        vocab_size_bytecode=len(bytecode_tokenizer.word_index) + 1
    )

    print("🚀 Training the model...")
    model.fit(
        [X_source_train, X_bytecode_train],
        y_train,
        validation_split=0.1,
        epochs=5,
        batch_size=32
    )

    print("📊 Evaluating the model...")
    y_pred = model.predict([X_source_test, X_bytecode_test])
    y_pred_labels = (y_pred > 0.5).astype(int)

    print("✅ Classification Report:")
    print(classification_report(y_test, y_pred_labels))

if __name__ == '__main__':
    main()
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, Bidirectional, LSTM, Dense, Concatenate

def load_contract_data(source_folder, bytecode_folder, labels_file):
    df = pd.read_csv(labels_file)
    df.columns = [col.strip().lower() for col in df.columns]

    if 'filename' not in df.columns:
        raise ValueError("CSV must contain a 'filename' column.")
    if 'reentrancy' not in df.columns:
        raise ValueError("CSV must contain a 'reentrancy' column for labels.")

    sources, bytecodes, labels = [], [], []

    for _, row in df.iterrows():
        fname = str(row['filename']).strip()
        label = int(row['reentrancy'])

        if not fname.endswith('.sol'):
            fname += '.sol'

        basename = os.path.splitext(os.path.basename(fname))[0]

        source_path = os.path.join(source_folder, fname)
        byte_path = os.path.join(bytecode_folder, f"{basename}.bin")

        if not os.path.isfile(source_path):
            print(f"⚠️ Missing source file: {source_path}")
            continue
        if not os.path.isfile(byte_path):
            print(f"⚠️ Missing bytecode file: {byte_path}")
            continue

        try:
            with open(source_path, 'r', encoding='utf-8') as f:
                source = f.read()

            with open(byte_path, 'r', encoding='utf-8') as f:
                bytecode_hex = f.read().strip()
                bytecode_tokens = ' '.join([bytecode_hex[i:i+2] for i in range(0, len(bytecode_hex), 2)])

            sources.append(source)
            bytecodes.append(bytecode_tokens)
            labels.append(label)
        except Exception as e:
            print(f"❌ Error reading {fname}: {e}")
            continue

    if len(sources) == 0:
        raise ValueError("No valid contracts loaded. Check file paths and naming consistency.")

    print(f"✅ Successfully loaded {len(sources)} contracts.")
    return sources, bytecodes, np.array(labels)

def tokenize_inputs(source_texts, bytecode_texts, max_words=10000, max_len=500):
    source_tokenizer = Tokenizer(num_words=max_words, oov_token="<OOV>")
    bytecode_tokenizer = Tokenizer(num_words=max_words, oov_token="<OOV>")

    source_tokenizer.fit_on_texts(source_texts)
    bytecode_tokenizer.fit_on_texts(bytecode_texts)

    source_seq = source_tokenizer.texts_to_sequences(source_texts)
    bytecode_seq = bytecode_tokenizer.texts_to_sequences(bytecode_texts)

    source_seq = pad_sequences(source_seq, maxlen=max_len, padding='post')
    bytecode_seq = pad_sequences(bytecode_seq, maxlen=max_len, padding='post')

    return source_seq, bytecode_seq, source_tokenizer, bytecode_tokenizer

def build_bilstm_model(vocab_size_source, vocab_size_bytecode, embedding_dim=64, input_len=500):
    source_input = Input(shape=(input_len,), name='source_input')
    bytecode_input = Input(shape=(input_len,), name='bytecode_input')

    source_embed = Embedding(vocab_size_source, embedding_dim)(source_input)
    bytecode_embed = Embedding(vocab_size_bytecode, embedding_dim)(bytecode_input)

    source_lstm = Bidirectional(LSTM(64))(source_embed)
    bytecode_lstm = Bidirectional(LSTM(64))(bytecode_embed)

    merged = Concatenate()([source_lstm, bytecode_lstm])
    dense = Dense(64, activation='relu')(merged)
    output = Dense(1, activation='sigmoid')(dense)

    model = Model(inputs=[source_input, bytecode_input], outputs=output)
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

def main():
    SOURCE_FOLDER = 'ESC_SOL'
    BYTECODE_FOLDER = 'ESC_BYTECODE'
    LABELS_FILE = 'labelled_ESC_1001.csv'

    print("🔄 Loading contracts and labels...")
    sources, bytecodes, labels = load_contract_data(SOURCE_FOLDER, BYTECODE_FOLDER, LABELS_FILE)

    print("🔄 Tokenizing and padding sequences...")
    source_seq, bytecode_seq, source_tokenizer, bytecode_tokenizer = tokenize_inputs(sources, bytecodes)

    print("🔄 Splitting into train/test...")
    X_source_train, X_source_test, X_bytecode_train, X_bytecode_test, y_train, y_test = train_test_split(
        source_seq, bytecode_seq, labels, test_size=0.2, random_state=42
    )

    print("🔄 Building the Bi-LSTM model...")
    model = build_bilstm_model(
        vocab_size_source=len(source_tokenizer.word_index) + 1,
        vocab_size_bytecode=len(bytecode_tokenizer.word_index) + 1
    )

    print("🚀 Training the model...")
    model.fit(
        [X_source_train, X_bytecode_train],
        y_train,
        validation_split=0.1,
        epochs=5,
        batch_size=32
    )

    print("📊 Evaluating the model...")
    y_pred = model.predict([X_source_test, X_bytecode_test])
    y_pred_labels = (y_pred > 0.5).astype(int)

    print("✅ Classification Report:")
    print(classification_report(y_test, y_pred_labels))

if __name__ == '__main__':
    main()
```

```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import torch
from torch.utils.data import Dataset
from transformers import RobertaTokenizer, RobertaForSequenceClassification, Trainer, TrainingArguments, DataCollatorWithPadding
from datasets import Dataset as HFDataset

# ---- Set this env var before importing transformers if TensorFlow is installed ----
os.environ["TRANSFORMERS_NO_TF"] = "1"


class CodeByteDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_length=512):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_length = max_length
    
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = self.texts[idx]
        label = self.labels[idx]
        encoding = self.tokenizer(text, truncation=True, padding='max_length', max_length=self.max_length, return_tensors='pt')
        # Squeeze to remove batch dimension 1
        item = {key: val.squeeze(0) for key, val in encoding.items()}
        item['labels'] = torch.tensor(label)
        return item


def load_data(source_folder, bytecode_folder, labels_csv):
    df = pd.read_csv(labels_csv)
    df.columns = [c.strip().lower() for c in df.columns]

    texts = []
    labels = []

    for _, row in df.iterrows():
        fname = str(row['filename']).strip()
        if not fname.endswith('.sol'):
            fname += '.sol'
        
        # Read source code
        source_path = os.path.join(source_folder, fname)
        if not os.path.exists(source_path):
            print(f"Warning: Source file missing: {source_path}")
            continue
        with open(source_path, 'r', encoding='utf-8', errors='ignore') as f:
            source_code = f.read()

        # Read bytecode text (assume .txt extension)
        bytecode_fname = fname.replace('.sol', '.txt')
        bytecode_path = os.path.join(bytecode_folder, bytecode_fname)
        bytecode_text = ""
        if os.path.exists(bytecode_path):
            with open(bytecode_path, 'r', encoding='utf-8', errors='ignore') as f:
                bytecode_text = f.read()
        else:
            print(f"Warning: Bytecode file missing: {bytecode_path}")

        # Combine source and bytecode with special token
        combined_text = source_code + "\n<BYTECODE> " + bytecode_text
        texts.append(combined_text)
        labels.append(int(row['reentrancy']))

    return texts, labels


def main():
    # Folders and files
    source_folder = './ESC_SOL'        # Your .sol files folder
    bytecode_folder = './ESC_BYTECODE'         # Your bytecode .txt files folder
    labels_csv = 'labelled_ESC_1001.csv'             # CSV file with filename, reentrancy

    # Load data
    texts, labels = load_data(source_folder, bytecode_folder, labels_csv)
    print(f"Loaded {len(texts)} samples")

    # Train test split
    train_texts, val_texts, train_labels, val_labels = train_test_split(texts, labels, test_size=0.2, random_state=42)

    # Tokenizer and model from CodeBERT (Roberta base)
    tokenizer = RobertaTokenizer.from_pretrained('microsoft/codebert-base')
    model = RobertaForSequenceClassification.from_pretrained('microsoft/codebert-base', num_labels=2)

    # Create datasets
    train_dataset = CodeByteDataset(train_texts, train_labels, tokenizer)
    val_dataset = CodeByteDataset(val_texts, val_labels, tokenizer)

    # Data collator for dynamic padding
    data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

    # Training arguments
    training_args = TrainingArguments(
        output_dir='./results',
        evaluation_strategy='epoch',
        save_strategy='epoch',
        learning_rate=2e-5,
        per_device_train_batch_size=8,
        per_device_eval_batch_size=8,
        num_train_epochs=3,
        weight_decay=0.01,
        load_best_model_at_end=True,
        metric_for_best_model='accuracy',
        logging_dir='./logs',
        logging_steps=10,
        save_total_limit=2,
        seed=42,
        push_to_hub=False,
    )

    # Define compute_metrics function
    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        preds = logits.argmax(-1)
        from sklearn.metrics import accuracy_score, precision_recall_fscore_support
        acc = accuracy_score(labels, preds)
        precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='binary')
        return {
            'accuracy': acc,
            'f1': f1,
            'precision': precision,
            'recall': recall
        }

    # Trainer
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        tokenizer=tokenizer,
        data_collator=data_collator,
        compute_metrics=compute_metrics,
    )

    # Train
    trainer.train()

    # Evaluate
    results = trainer.evaluate()
    print(results)

    # Predict on val set and show classification report
    preds_output = trainer.predict(val_dataset)
    preds = preds_output.predictions.argmax(-1)
    print(classification_report(val_labels, preds, digits=4))


if __name__ == "__main__":
    main()
```

```python
import os
import zipfile

def zip_folder(folder_path, output_zip_path):
    with zipfile.ZipFile(output_zip_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for root, _, files in os.walk(folder_path):
            for file in files:
                abs_path = os.path.join(root, file)
                rel_path = os.path.relpath(abs_path, folder_path)
                zipf.write(abs_path, arcname=rel_path)

# Example usage:
zip_folder('ESC_VUL', 'ESC_VUL.zip')
```

```python
import os
import shutil

def extract_matching_folders_with_logs(folder1, folder2, folder3):
    # Step 1: Collect all contract names (without .sol)
    contract_names = set()
    for file in os.listdir(folder1):
        if file.endswith('.sol'):
            contract_names.add(os.path.splitext(file)[0].lower())  # case-insensitive match

    # Step 2: Recursively search folder2 for matching folder names
    matched_paths = []
    for root, dirs, _ in os.walk(folder2):
        for d in dirs:
            if d.lower() in contract_names:
                matched_paths.append(os.path.join(root, d))

    # Step 3: Create destination folder3
    os.makedirs(folder3, exist_ok=True)

    # Step 4: Copy matching folders (with contents) to folder3
    for src_folder in matched_paths:
        folder_name = os.path.basename(src_folder)
        dst_folder = os.path.join(folder3, folder_name)
        if os.path.exists(dst_folder):
            print(f"Skipping (already exists): {dst_folder}")
        else:
            shutil.copytree(src_folder, dst_folder)
            print(f"Copied: {src_folder} -> {dst_folder}")

    print("✅ Done copying all matching folders with contents (including logs).")

# Example usage
folder1 = "./ESC_SOL"  # Folder with .sol files
folder2 = "../smartbugs/results/slither-0.10.4"  # Search recursively inside this
folder3 = "./folder3"  # Copy matched folders here

extract_matching_folders_with_logs(folder1, folder2, folder3)
```
